# Задача 1
Подсчёт гласных букв (a, e, i, o, u) в строке.

```python
def count_vowels(s):
    vowels = 'aeiouAEIOU'
    return sum(1 for ch in s if ch in vowels)

# Тесты с выводом
tests = [
    ("hello", 2),
    ("why", 0),
    ("AEIOU", 5),
    ("bcdfg", 0),
    ("education", 5)
]

for i, (s, exp) in enumerate(tests, 1):
    res = count_vowels(s)
    ok = res == exp
    print(f"Тест {i}: {'OK' if ok else 'FAIL'} | count_vowels('{s}') = {res} (ожидалось {exp})")
print("Задача 1: завершено")
```

# Задача 2
Проверка, что каждый символ встречается ровно один раз.

```python
def all_unique(s):
    return len(s) == len(set(s))

tests = [
    ("abc", True),
    ("aabb", False),
    ("", True),
    ("abcdA", True),
    ("hello", False)
]

for i, (s, exp) in enumerate(tests, 1):
    res = all_unique(s)
    ok = res == exp
    print(f"Тест {i}: {'OK' if ok else 'FAIL'} | all_unique('{s}') = {res} (ожидалось {exp})")
print("Задача 2: завершено")
```

# Задача 3
Количество единичных битов в положительном числе.

```python
def count_bits(n):
    return bin(n).count('1')

tests = [
    (0, 0),
    (1, 1),
    (7, 3),
    (8, 1),
    (255, 8)
]

for i, (n, exp) in enumerate(tests, 1):
    res = count_bits(n)
    ok = res == exp
    print(f"Тест {i}: {'OK' if ok else 'FAIL'} | count_bits({n}) = {res} (ожидалось {exp})")
print("Задача 3: завершено")
```

# Задача 4
Сколько раз нужно перемножать цифры, чтобы получить однозначное число.

```python
def magic(n):
    if n < 10:
        return 0
    cnt = 0
    while n >= 10:
        prod = 1
        while n:
            prod *= n % 10
            n //= 10
        n = prod
        cnt += 1
    return cnt

tests = [
    (39, 3),
    (4, 0),
    (999, 4),
    (25, 2),
    (10, 1)
]

for i, (n, exp) in enumerate(tests, 1):
    res = magic(n)
    ok = res == exp
    print(f"Тест {i}: {'OK' if ok else 'FAIL'} | magic({n}) = {res} (ожидалось {exp})")
print("Задача 4: завершено")
```

# Задача 5  
**Mean Squared Error** двух векторов одинаковой длины.

```python
def mse(pred, true):
    if len(pred) != len(true):
        raise ValueError("Векторы должны быть одинаковой длины")
    n = len(pred)
    return sum((p - t) ** 2 for p, t in zip(pred, true)) / n

tests = [
    ([1, 2, 3], [1, 2, 3], 0.0),
    ([1, 2, 3], [2, 4, 6], 14/3),
    ([0, 0], [1, 1], 1.0)
]

for i, (pred, true, exp) in enumerate(tests, 1):
    res = mse(pred, true)
    ok = abs(res - exp) < 1e-9
    print(f"Тест {i}: {'OK' if ok else 'FAIL'} | mse({pred}, {true}) = {res:.6f} (ожидалось {exp})")
print("Задача 5: завершено")
```

# Задача 6
Разложение числа на простые множители в виде `(p1**e1)(p2**e2)…`.

```python
def prime_factors(n):
    factors = []
    
    cnt = 0
    while n % 2 == 0:
        cnt += 1
        n //= 2
    if cnt:
        factors.append(f"(2**{cnt})" if cnt > 1 else "(2)")
    # нечётные
    i = 3
    while i * i <= n:
        cnt = 0
        while n % i == 0:
            cnt += 1
            n //= i
        if cnt:
            factors.append(f"({i}**{cnt})" if cnt > 1 else f"({i})")
        i += 2
    if n > 1:
        factors.append(f"({n})")
    return ''.join(factors)

tests = [
    (86240, "(2**5)(5)(7**2)(11)"),
    (12, "(2**2)(3)"),
    (17, "(17)"),
    (100, "(2**2)(5**2)")
]

for i, (n, exp) in enumerate(tests, 1):
    res = prime_factors(n)
    ok = res == exp
    print(f"Тест {i}: {'OK' if ok else 'FAIL'} | prime_factors({n}) = '{res}' (ожидалось '{exp}')")
print("Задача 6: завершено")
```

# Задача 7  
Можно ли построить пирамиду из `n` кубиков? (сумма квадратов 1²+…+k²).

```python
def pyramid(number):
    total = 0
    k = 0
    while total < number:
        k += 1
        total += k * k
        if total == number:
            return k
    return "нет нельзя"

tests = [
    (1, 1),
    (5, 2),
    (14, 3),
    (30, 4),               # 1+4+9+16 = 30
    (2, "нет нельзя")
]

for i, (n, exp) in enumerate(tests, 1):
    res = pyramid(n)
    ok = res == exp
    print(f"Тест {i}: {'OK' if ok else 'FAIL'} | pyramid({n}) = {res} (ожидалось {exp})")
print("Задача 7: завершено")
```

# Задача 8  
Сбалансированное число – сумма цифр слева от середины равна сумме справа.

```python
def is_balanced(n):
    s = str(n)
    length = len(s)
    if length <= 2:
        return True
    mid = length // 2
    left_end = mid - 1 if length % 2 == 0 else mid
    left_sum = sum(int(d) for d in s[:left_end])
    right_start = mid + 1 if length % 2 == 0 else mid + 1
    right_sum = sum(int(d) for d in s[right_start:])
    return left_sum == right_sum

tests = [
    (12321, True),   # 1+2 == 2+1
    (123321, True),  # 1+2+3 == 3+2+1
    (123, False),    # 1 != 3 (средняя 2)
    (1234, False),   # 1+2=3, 3+4=7
    (9980, False),   # 9+9=18, 8+0=8
    (1, True),
    (11, True)
]

for i, (n, exp) in enumerate(tests, 1):
    res = is_balanced(n)
    ok = res == exp
    print(f"Тест {i}: {'OK' if ok else 'FAIL'} | is_balanced({n}) = {res} (ожидалось {exp})")
print("Задача 8: завершено")
```
